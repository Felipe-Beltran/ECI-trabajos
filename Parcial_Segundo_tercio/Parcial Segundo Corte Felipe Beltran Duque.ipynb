### 5) Análisis de complejidad **[15 pts]**
1. Representación y Agregar Conexión

Mejor Caso (O(1)): Ocurre cuando la estación de origen ya existe en el diccionario del grafo y la conexión que se desea agregar ya está registrada o se inserta al final de la lista de adyacencia existente. La inserción en listas de Python toma tiempo constante amortizado.

Peor Caso (O(V)): Ocurre cuando la estación es completamente nueva y requiere crear la clave en el diccionario, o bien cuando se verifica la existencia de una arista y la lista de vecinos de la estación es muy grande (en grafos densos, proporcional al número de vértices V).


2. Recorrido BFS (Búsqueda en Anchura / Niveles)

Mejor Caso (O(V + E)): En grafos, BFS siempre debe evaluar al menos el nodo inicial y sus conexiones directas para empezar a formar niveles. El mejor caso se da cuando el nodo inicial está altamente conectado con el resto de la red de manera equilibrada, permitiendo que la cola procese los vértices sin desbalances profundos. 

(Nota: Teóricamente, en teoría de grafos, BFS y DFS operan en tiempo acotado por O(V + E) ya que visitan todos los vértices y aristas alcanzables).
Peor Caso (O(V + E)): Ocurre en grafos con una estructura lineal (forma de cadena o "liana", ej. ⁠Central -> Norte -> Parque -> ...⁠), donde el algoritmo se ve obligado a procesar cada vértice uno por uno de forma secuencial, llenando y vaciando la cola en un solo hilo de profundidad máxima.


3. Recorrido DFS (Búsqueda en Profundidad)

Mejor Caso (O(V + E)): Cuando el nodo inicial encuentra de inmediato una ruta lineal profunda y ramificada de manera uniforme, permitiendo que la pila de recursividad avance sin retrocesos excesivos (backtracking).

Peor Caso (O(V + E)): Se presenta en grafos altamente ramificados o con estructuras complejas donde el algoritmo explora caminos erróneos y debe realizar un proceso exhaustivo de backtracking (retroceso) por todas las ramas muertas antes de completar la visita de la red. En términos espaciales, el peor caso de la pila recursiva alcanza O(V) debido a la profundidad de la red.

4. Búsqueda de Ruta Más Corta (BFS Óptimo)

Mejor Caso (O(1)): Ocurre cuando la estación de origen es exactamente igual a la de destino (⁠origen == destino⁠). El algoritmo detecta la coincidencia de inmediato en la primera iteración sin necesidad de recorrer vecinos ni vaciar la cola.

Peor Caso (O(V + E)): Ocurre cuando la estación de destino se encuentra en el extremo más lejano de la red (por ejemplo, el nivel más profundo de la red, como de ⁠Central⁠ a ⁠Terminal⁠) o cuando la ruta no existe y el algoritmo se ve obligado a explorar absolutamente todos los vértices y aristas accesibles antes de concluir la búsqueda.

```python
class RedTransporte:
    def __init__(self):
        self.grafo = {}
 
    def agregar_conexion(self, origen, destino):
        """Agrega una arista no dirigida y sin duplicados[span_0](start_span)[span_0](end_span)."""
        if origen not in self.grafo:
            self.grafo[origen] = []
        if destino not in self.grafo:
            self.grafo[destino] = []
        if destino not in self.grafo[origen]:
            self.grafo[origen].append(destino)
        if origen not in self.grafo[destino]:
            self.grafo[destino].append(origen)
 
    def mostrar_red(self):
        """Muestra cada estación y sus conexiones directas[span_1](start_span)[span_1](end_span)."""
        print("\n--- RED DE ESTACIONES ---")
        for estacion, vecinos in self.grafo.items():
            print(f"{estacion}: {', '.join(vecinos)}")
 
    def bfs(self, inicio):
        """Recorrido BFS usando una cola[span_2](start_span)[span_2](end_span)."""
        if inicio not in self.grafo:
            return None, None
        visitados = set([inicio])
        cola = [inicio]
        orden = []
        niveles = {inicio: 0}
       
        while cola:
            actual = cola.pop(0)
            orden.append(actual)
            for vecino in self.grafo[actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    niveles[vecino] = niveles[actual] + 1
                    cola.append(vecino)
        return orden, niveles
 
    def dfs(self, inicio):
        """Recorrido DFS usando una pila[span_3](start_span)[span_3](end_span)."""
        if inicio not in self.grafo:
            return None
        visitados = set()
        pila = [inicio]
        orden = []
       
        while pila:
            actual = pila.pop()
            if actual not in visitados:
                visitados.add(actual)
                orden.append(actual)
                for vecino in reversed(self.grafo[actual]):
                    if vecino not in visitados:
                        pila.append(vecino)
        return orden
 
    def ruta_minima(self, origen, destino):
        """Encuentra la ruta con menor cantidad de conexiones usando BFS[span_4](start_span)[span_4](end_span)."""
        if origen not in self.grafo or destino not in self.grafo:
            return None, 0
        visitados = set([origen])
        cola = [[origen]]
       
        while cola:
            ruta = cola.pop(0)
            actual = ruta[-1]
            if actual == destino:
                return ruta, len(ruta) - 1
            for vecino in self.grafo[actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    nueva_ruta = list(ruta)
                    nueva_ruta.append(vecino)
                    cola.append(nueva_ruta)
        return None, 0
 
def menu():
    red = RedTransporte()
   
    # Cargar red inicial del parcial[span_5](start_span)[span_5](end_span)
    iniciales = [
        ("Central", "Norte"), ("Central", "Museo"),
        ("Norte", "Parque"), ("Norte", "Universidad"),
        ("Museo", "Hospital"), ("Universidad", "Terminal"),
        ("Hospital", "Terminal")
    ]
    for o, d in iniciales:
        red.agregar_conexion(o, d)
 
    while True:
        print("\n================ MENÚ RED DE TRANSPORTE ================")
        print("1. Mostrar red de estaciones")
        print("2. Ejecutar BFS y DFS desde una estación")
        print("3. Encontrar ruta mínima entre dos estaciones")
        print("4. Agregar nueva estación (Ej. Aeropuerto)")
        print("5. Salir")
       
        op = input("Elige una opción (1-5): ").strip()
       
        if op == "1":
            red.mostrar_red()
           
        elif op == "2":
            est = input("Ingrese la estación inicial: ").strip()
            if est not in red.grafo:
                print("\n[Error]: La estación no existe en la red[span_6](start_span)[span_6](end_span).")
            else:
                orden_b, niv = red.bfs(est)
                orden_d = red.dfs(est)
                print(f"\nBFS desde {est}: {' -> '.join(orden_b)}")
                for n in sorted(set(niv.values())):
                    est_n = [e for e, v in niv.items() if v == n]
                    print(f"  Nivel {n}: {', '.join(est_n)}")
                print(f"DFS desde {est}: {' -> '.join(orden_d)}")
               
        elif op == "3":
            orig = input("Estación de origen: ").strip()
            dest = input("Estación de destino: ").strip()
            ruta, conns = red.ruta_minima(orig, dest)
            if ruta:
                print(f"\nRuta encontrada:\n{' -> '.join(ruta)}")
                print(f"Cantidad de conexiones: {conns}")
            else:
                print("\n[Error]: No existe ruta o alguna estación no es válida[span_7](start_span)[span_7](end_span).")
               
        elif op == "4":
            nueva = input("Nombre de la nueva estación (Ej: Aeropuerto): ").strip()
            if not nueva:
                print("\n[Error]: El nombre no puede estar vacío.")
                continue
            print(f"Estaciones disponibles: {list(red.grafo.keys())}")
            c1 = input("Conectar con estación 1 (Ej: Terminal): ").strip()
            c2 = input("Conectar con estación 2 (Ej: Hospital): ").strip()
           
            if c1 in red.grafo and c2 in red.grafo:
                red.agregar_conexion(nueva, c1)
                red.agregar_conexion(nueva, c2)
                print(f"\n[Éxito]: Estación '{nueva}' agregada correctamente[span_8](start_span)[span_8](end_span).")
                red.mostrar_red()
            else:
                print("\n[Error]: Una o ambas estaciones de conexión no existen.")
               
        elif op == "5":
            print("\nSaliendo del sistema...")
            break
        else:
            print("\n[Error]: Opción inválida, digite un número entre 1 y 5.")
 
if __name__ == "__main__":
    menu()
```
