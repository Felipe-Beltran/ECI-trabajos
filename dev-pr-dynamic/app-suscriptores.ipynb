## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python

def streaming(n, A):
   
    if n < len(A) and A[n] is not None:
        return A[n]
    
   
    if n == 0:
        resultado = 0
    elif n == 1:
        resultado = 1
    else:
        
        resultado = streaming(n-1, A) + streaming(n-2, A)
    
    
    while len(A) <= n:
        A.append(None)
        
  
    A[n] = resultado
    return resultado



def main():
    decision = "si"
   
    A = [0, 1] 
    
    while decision == "si":
        n = int(input("Cual es el mes que desea generar, ponga un numero entero: "))
       
        respuesta = streaming(n, A)
        
        print("La respuesta es:", respuesta)
        print("Memoria actual de meses calculados (A):", A)
        
        decision = input("Hay otro Usuario? En dado caso que sí, responder únicamente con 'si': ").strip().lower()

main()
```
