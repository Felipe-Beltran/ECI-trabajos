## 7) Análisis de complejidad **[5 pts]**
1. Creación de la Tabla Hash
Consiste en crear una tabla de tamaño m, indicada por el usuario, donde inicialmente todas las posiciones están vacías.
Mejor caso: O(m). Se inicializan todas las posiciones de la tabla sin inconvenientes.
Peor caso: O(m). Se deben crear igualmente las m posiciones, independientemente de su contenido.

2. Insertar productos
Consiste en calcular la posición mediante la función hash y almacenar el producto. Si existe una colisión, se agrega a la lista correspondiente.
Mejor caso: O(1). El producto se inserta directamente en una posición vacía, sin colisiones.
Peor caso: O(n). Todos los productos llegan a la misma posición y se debe recorrer la lista para verificar que el código no esté repetido antes de insertarlo.

3. Manejo de colisiones
Se utiliza encadenamiento, donde cada posición puede almacenar varios productos mediante una lista.
Mejor caso: O(1). No existen colisiones y cada posición contiene como máximo un producto.
Peor caso: O(n). Todos los productos generan el mismo índice y quedan almacenados en una sola lista.

4. Buscar un producto
Se calcula la posición con la función hash y se busca el código dentro de la lista correspondiente, sin recorrer toda la tabla.
Mejor caso: O(1). El producto se encuentra en la primera posición de la lista.
Peor caso: O(n). Todos los productos están en la misma posición y el producto buscado está al final de la lista o no existe.

5. Eliminar un producto
Se calcula su posición mediante la función hash y se elimina únicamente de la lista correspondiente.
Mejor caso: O(1). El producto está al inicio de la lista y se elimina directamente.
Peor caso: O(n). Todos los productos están en la misma posición y es necesario recorrer la lista hasta encontrar el producto o comprobar que no existe.

6. Actualizar inventario
Se busca el producto utilizando su código y se modifica la cantidad disponible.
Mejor caso: O(1). El producto se encuentra directamente en la primera posición de la lista y se actualiza su cantidad.
Peor caso: O(n). Todos los productos generan colisiones y el producto que se desea actualizar está al final de la lista o no existe.

7. Mostrar la Tabla Hash
Consiste en recorrer todas las posiciones de la tabla y mostrar los productos almacenados en cada una.
Mejor caso: O(m + n). Se recorren las m posiciones y se muestran los n productos distribuidos en ellas.
Peor caso: O(m + n). Se deben recorrer todas las posiciones y mostrar todos los productos, incluso cuando existen muchas colisiones






## 8) Código funcional **[10 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python

class TablaHash:
    def __init__(self, tamano):
        if tamano <= 0:
            raise ValueError("El tamaño debe ser mayor que cero.")

        self.tamano = tamano

        
        self.tabla = [[] for _ in range(tamano)]

    def funcion_hash(self, codigo):
        return codigo % self.tamano

    def buscar_indice(self, codigo):
        
        indice = self.funcion_hash(codigo)

        
        for producto in self.tabla[indice]:
            if producto["codigo"] == codigo:
                return producto

        return None

    def insertar(self, codigo, nombre, precio, cantidad):
        if codigo < 0 or precio < 0 or cantidad < 0:
            print("El código, precio y cantidad no pueden ser negativos.")
            return False

        if self.buscar_indice(codigo) is not None:
            print("Ya existe un producto con ese código.")
            return False

        indice = self.funcion_hash(codigo)

        producto = {
            "codigo": codigo,
            "nombre": nombre,
            "precio": precio,
            "cantidad": cantidad
        }

       
        self.tabla[indice].append(producto)

        print(f"Producto insertado en el índice {indice}.")
        return True

    def buscar(self, codigo):
        producto = self.buscar_indice(codigo)

        if producto is None:
            print("El producto no está registrado.")
            return None

        print("\nProducto encontrado")
        print(f"Código: {producto['codigo']}")
        print(f"Nombre: {producto['nombre']}")
        print(f"Precio: ${producto['precio']:,.0f}")
        print(f"Cantidad disponible: {producto['cantidad']}")

        return producto

    def eliminar(self, codigo):
        indice = self.funcion_hash(codigo)

        for posicion, producto in enumerate(self.tabla[indice]):
            if producto["codigo"] == codigo:
                eliminado = self.tabla[indice].pop(posicion)
                print(f"Producto '{eliminado['nombre']}' eliminado.")
                return True

        print("No se encontró el producto.")
        return False

    def actualizar_cantidad(self, codigo, nueva_cantidad):
        if nueva_cantidad < 0:
            print("La cantidad no puede ser negativa.")
            return False

        producto = self.buscar_indice(codigo)

        if producto is None:
            print("El producto no está registrado.")
            return False

        producto["cantidad"] = nueva_cantidad

        print("Inventario actualizado correctamente.")
        print(f"{producto['nombre']} - {nueva_cantidad} unidades")
        return True

    def mostrar(self):
        print("\n" + "=" * 45)
        print("             TABLA HASH")
        print("=" * 45)

        for indice, cadena in enumerate(self.tabla):
            if not cadena:
                print(f"{indice} -> Vacío")
            else:
                print(f"{indice} -> ", end="")

                for posicion, producto in enumerate(cadena):
                    if posicion > 0:
                        print("       ", end="")

                    print(
                        f"{producto['codigo']} - {producto['nombre']} "
                        f"(${producto['precio']:,.0f}, "
                        f"{producto['cantidad']} unidades)"
                    )

        print("=" * 45)


def leer_entero(mensaje, minimo=None):
    while True:
        try:
            valor = int(input(mensaje))

            if minimo is not None and valor < minimo:
                print(f"El valor debe ser mayor o igual a {minimo}.")
                continue

            return valor

        except ValueError:
            print("Ingresa un número entero válido.")


def leer_precio(mensaje):
    while True:
        try:
            precio = float(input(mensaje).replace(",", "."))

            if precio < 0:
                print("El precio no puede ser negativo.")
                continue

            return precio

        except ValueError:
            print("Ingresa un precio válido.")


def menu():
    print("\n===== INVENTARIO DE PRODUCTOS =====")
    print("1. Insertar productos")
    print("2. Buscar producto")
    print("3. Eliminar producto")
    print("4. Actualizar cantidad")
    print("5. Mostrar Tabla Hash")
    print("6. Salir")


def ejecutar():
    print("CREACIÓN DE LA TABLA HASH")
    tamano = leer_entero("Ingrese el tamaño de la tabla hash: ", 1)
    inventario = TablaHash(tamano)

    while True:
        menu()
        opcion = leer_entero("Seleccione una opción: ", 1)

        if opcion == 1:
            cantidad_productos = leer_entero(
                "¿Cuántos productos desea registrar?: ", 0
            )

            for _ in range(cantidad_productos):
                print("\nIngrese los datos del producto:")

                codigo = leer_entero("Código: ", 0)
                nombre = input("Nombre: ").strip()

                while not nombre:
                    print("El nombre no puede estar vacío.")
                    nombre = input("Nombre: ").strip()

                precio = leer_precio("Precio: $")
                cantidad = leer_entero("Cantidad disponible: ", 0)

                inventario.insertar(codigo, nombre, precio, cantidad)

        elif opcion == 2:
            codigo = leer_entero("Código del producto que desea buscar: ", 0)
            inventario.buscar(codigo)

        elif opcion == 3:
            codigo = leer_entero("Código del producto que desea eliminar: ", 0)
            inventario.eliminar(codigo)

        elif opcion == 4:
            codigo = leer_entero("Código del producto: ", 0)
            nueva_cantidad = leer_entero("Nueva cantidad: ", 0)
            inventario.actualizar_cantidad(codigo, nueva_cantidad)

        elif opcion == 5:
            inventario.mostrar()

        elif opcion == 6:
            print("Programa finalizado.")
            break

        else:
            print("Opción no válida. Selecciona una opción del 1 al 6.")


# Ejecutar el programa
ejecutar()
```

#### Código base para benchmarking

```python


import time
import matplotlib.pyplot as plt


def crear_tabla_para_prueba(cantidad_productos):
    """
    Crea una Tabla Hash y agrega una cantidad determinada
    de productos para realizar pruebas de rendimiento.
    """


    tabla = TablaHash(cantidad_productos * 2 + 1)

    for codigo in range(cantidad_productos):
        indice = tabla.funcion_hash(codigo)

        producto = {
            "codigo": codigo,
            "nombre": f"Producto {codigo}",
            "precio": 1000,
            "cantidad": 10
        }

        tabla.tabla[indice].append(producto)

    return tabla


def medir_busqueda(cantidad_productos, repeticiones=1000):
    """
    Mide el tiempo promedio necesario para buscar
    un producto en la Tabla Hash.
    """

    tabla = crear_tabla_para_prueba(cantidad_productos)

    codigo_buscar = cantidad_productos - 1

    inicio = time.perf_counter()

    for _ in range(repeticiones):
        tabla.buscar_indice(codigo_buscar)

    fin = time.perf_counter()

    tiempo_promedio = (
        (fin - inicio) / repeticiones
    ) * 1000

    return tiempo_promedio


tamanos = [100, 1000, 5000, 10000, 20000]

tiempos = []

print("PRUEBAS DE DESEMPEÑO\n")

for cantidad in tamanos:
    tiempo = medir_busqueda(cantidad)
    tiempos.append(tiempo)

    print(
        f"{cantidad} productos -> "
        f"{tiempo:.8f} ms por búsqueda"
    )


plt.figure(figsize=(8, 5))

plt.plot(tamanos, tiempos, marker="o")

plt.title("Tiempo de búsqueda en la Tabla Hash")
plt.xlabel("Cantidad de productos")
plt.ylabel("Tiempo promedio de búsqueda (ms)")
plt.grid()

plt.show()
```
