### 5) Análisis de complejidad **[15 pts]**

Nuestro programa genera todas las conexiones posibles entre los componentes. Como cada componente puede conectarse con los demás, la cantidad de aristas es:

Variables:

V: cantidad de vértices o componentes.


E: cantidad de aristas o conexiones posibles.

E = V(V - 1) / 2


Por lo tanto, E = O(V²).
Después, el algoritmo de Kruskal ordena las conexiones de menor a mayor peso, lo que tiene una complejidad de O(E log E).

Al reemplazar E por V², obtenemos:

Complejidad final: O(V² log V).

Complejidad espacial: O(V²)


Se debe a que almacenamos todas las conexiones posibles en una lista antes de ordenarlas y seleccionar las que pertenecen al árbol de expansión mínima.


Mejor caso: O(V² log V)
Se presenta cuando las conexiones seleccionadas permiten completar el árbol rápidamente y se necesitan pocas revisiones para descartar ciclos. Sin embargo, el programa debe generar y ordenar todas las conexiones antes de ejecutar Kruskal.


Peor caso: O(V² log V)
Se presenta cuando es necesario revisar muchas conexiones, debido a que algunas forman ciclos y deben descartarse antes de completar el árbol de expansión mínima.



## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
import math
import matplotlib.pyplot as plt


class ArbolExpansionMinima:

    def __init__(self, componentes):
        self.componentes = componentes

    # Calcula la distancia entre dos componentes
    def calcular_distancia(self, posicion1, posicion2):
        x1, y1 = posicion1
        x2, y2 = posicion2

        distancia = math.sqrt(
            (x2 - x1) ** 2 + (y2 - y1) ** 2
        )

        return round(distancia, 2)

    # Genera todas las posibles conexiones
    def generar_conexiones(self):
        conexiones = []
        nombres = list(self.componentes.keys())

        for i in range(len(nombres)):
            for j in range(i + 1, len(nombres)):

                componente1 = nombres[i]
                componente2 = nombres[j]

                distancia = self.calcular_distancia(
                    self.componentes[componente1],
                    self.componentes[componente2]
                )

                conexiones.append(
                    (distancia, componente1, componente2)
                )

        return conexiones

    # Busca el grupo al que pertenece un componente
    def buscar(self, padre, componente):
        if padre[componente] != componente:
            padre[componente] = self.buscar(
                padre,
                padre[componente]
            )

        return padre[componente]

    # Une dos grupos diferentes
    def unir(self, padre, rango, componente1, componente2):

        raiz1 = self.buscar(padre, componente1)
        raiz2 = self.buscar(padre, componente2)

        if rango[raiz1] < rango[raiz2]:
            padre[raiz1] = raiz2

        elif rango[raiz1] > rango[raiz2]:
            padre[raiz2] = raiz1

        else:
            padre[raiz2] = raiz1
            rango[raiz1] += 1

    # Algoritmo de Kruskal
    def kruskal(self):

        # Generar todas las conexiones
        conexiones = self.generar_conexiones()

        # Ordenar de menor a mayor distancia
        conexiones.sort()

        padre = {}
        rango = {}

        # Cada componente comienza en su propio grupo
        for componente in self.componentes:
            padre[componente] = componente
            rango[componente] = 0

        mst = []
        longitud_total = 0

        # Seleccionar conexiones evitando ciclos
        for distancia, componente1, componente2 in conexiones:

            raiz1 = self.buscar(padre, componente1)
            raiz2 = self.buscar(padre, componente2)

            # Solo se agrega si no forma un ciclo
            if raiz1 != raiz2:

                mst.append(
                    (componente1, componente2, distancia)
                )

                longitud_total += distancia

                # Unir los grupos
                self.unir(
                    padre,
                    rango,
                    componente1,
                    componente2
                )

                # Un árbol con n componentes tiene n - 1 conexiones
                if len(mst) == len(self.componentes) - 1:
                    break

        return mst, round(longitud_total, 2)

    # Muestra la protoboard gráficamente
    def mostrar_protoboard(self, conexiones=None, titulo="Protoboard"):

        plt.figure(figsize=(10, 8))

        # Dibujar componentes
        for nombre, posicion in self.componentes.items():

            x, y = posicion

            plt.scatter(x, y, s=200)
            plt.text(
                x + 0.2,
                y + 0.2,
                nombre,
                fontsize=12
            )

        # Dibujar las conexiones seleccionadas
        if conexiones is not None:

            for componente1, componente2, distancia in conexiones:

                x1, y1 = self.componentes[componente1]
                x2, y2 = self.componentes[componente2]

                # Dibujar el cable
                plt.plot([x1, x2], [y1, y2])

                # Mostrar la distancia
                medio_x = (x1 + x2) / 2
                medio_y = (y1 + y2) / 2

                plt.text(
                    medio_x,
                    medio_y,
                    str(distancia),
                    fontsize=10
                )

        plt.title(titulo)
        plt.xlabel("Columnas")
        plt.ylabel("Filas")
        plt.grid(True)
        plt.show()


# ==========================================
# POSICIONES DE LOS COMPONENTES
# ==========================================

componentes = {
    "A": (2, 3),
    "B": (6, 5),
    "C": (10, 2),
    "D": (3, 10),
    "E": (8, 9),
    "F": (12, 7),
    "Buzzer": (8, 6)
}


# ==========================================
# MST ORIGINAL
# ==========================================

arbol = ArbolExpansionMinima(componentes)

# Mostrar distribución original
arbol.mostrar_protoboard(
    titulo="Distribución original"
)

# Ejecutar Kruskal
mst_original, longitud_original = arbol.kruskal()

# Mostrar resultado
print("========== RESULTADO MST ORIGINAL ==========")

for i, conexion in enumerate(mst_original, start=1):

    componente1, componente2, distancia = conexion

    print(
        f"Cable {i}: "
        f"{componente1} - {componente2} "
        f"= {distancia} unidades"
    )

print()
print(f"Longitud total de cable: {longitud_original} unidades")


# Mostrar conexiones del MST
arbol.mostrar_protoboard(
    mst_original,
    "Árbol de Expansión Mínima Original"
)


# ==========================================
# CAMBIO DE POSICIÓN DEL BUZZER
# ==========================================

# Nueva posición: J20
componentes["Buzzer"] = (10, 20)

# Crear nuevamente el árbol con la nueva posición
arbol_nuevo = ArbolExpansionMinima(componentes)

# Ejecutar nuevamente Kruskal
mst_nuevo, longitud_nueva = arbol_nuevo.kruskal()

print()
print("========== MST DESPUÉS DEL CAMBIO ==========")

for i, conexion in enumerate(mst_nuevo, start=1):

    componente1, componente2, distancia = conexion

    print(
        f"Cable {i}: "
        f"{componente1} - {componente2} "
        f"= {distancia} unidades"
    )

print()
print(f"Longitud total de cable: {longitud_nueva} unidades")


# Mostrar nueva protoboard
arbol_nuevo.mostrar_protoboard(
    mst_nuevo,
    "Árbol de Expansión Mínima después del cambio"
)


# ==========================================
# COMPARACIÓN FINAL
# ==========================================

print()
print("========== COMPARACIÓN ==========")

print(f"MST original - Longitud total: {longitud_original}")
print(f"MST después del cambio - Longitud total: {longitud_nueva}")

print()

if longitud_nueva > longitud_original:
    print("La cantidad total de cable AUMENTÓ.")

elif longitud_nueva < longitud_original:
    print("La cantidad total de cable DISMINUYÓ.")

else:
    print("La cantidad total de cable se mantuvo igual.")
```
