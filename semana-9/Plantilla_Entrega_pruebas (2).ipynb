## 1) Descripción del problema **[1 pts]**
Descripción del problema

Una empresa de mensajería cuenta con varios centros de distribución conectados mediante rutas terrestres. Para representar esta red se utilizará un grafo no dirigido, donde cada centro es un vértice y cada ruta entre dos centros es una arista.


Por ejemplo, el centro A (Centro Principal) está conectado con B (Centro Norte) y C (Centro Sur). A su vez, B está conectado con D y E, mientras que C está conectado con F.

El programa deberá permitir realizar dos tipos de recorridos:


* BFS (búsqueda en anchura): recorre primero los centros más cercanos al punto de inicio y luego avanza hacia los siguientes niveles. Por ejemplo, comenzando en A: A → B → C → D → E → F → G. En este caso, B y C están en el nivel 1, D, E y F en el nivel 2, y G en el nivel 3.


* DFS (búsqueda en profundidad): recorre una ruta lo más lejos posible antes de regresar y explorar otra. Por ejemplo, comenzando en A: A → B → D → E → G → F → C.

Además, el usuario podrá seleccionar el centro desde el cual desea comenzar, buscar un centro específico y conocer la ruta utilizada para llegar hasta él. Por ejemplo, desde A hasta F, una ruta encontrada mediante BFS sería A → C → F, con 2 conexiones.

Finalmente, el programa permitirá agregar nuevos centros y conexiones. Por ejemplo, se puede agregar el centro H y conectarlo con D y G, actualizando la lista de adyacencia y los recorridos de la red.

## 7) Análisis de complejidad **[5 pts]**

BFS — Breadth-First Search

El algoritmo BFS recorre los centros de la red por niveles, utilizando una cola para controlar el orden de visita. En el recorrido, cada centro puede ser visitado una vez y se revisan sus conexiones para determinar cuáles centros todavía no han sido visitados. Como el grafo se representa mediante una lista de adyacencia, el algoritmo puede recorrer los V centros y las E conexiones, por lo que su complejidad temporal es O(V + E) tanto en el mejor como en el peor caso. La diferencia entre los casos depende de cuándo se encuentre el centro buscado, pero para analizar el recorrido completo de la red se considera O(V + E).

BFS — Breadth-First Search

Mejor caso:
O(V + E)

Peor caso:
O(V + E)



DFS — Depth-First Search


El algoritmo DFS recorre la red profundizando en cada conexión antes de regresar para explorar otras alternativas. Para realizar el recorrido utiliza una estructura tipo pila, ya sea implementada directamente o mediante recursividad. Al utilizar una lista de adyacencia, cada centro se visita una vez y se revisan sus conexiones, por lo que se recorren V vértices y E aristas. Por esta razón, la complejidad temporal de DFS es O(V + E) tanto en el mejor como en el peor caso.

Mejor caso:
O(V + E)

Peor caso:
O(V + E)


## 8) Código funcional **[10 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
from collections import deque


class RedDistribucion:
    def __init__(self):
        self.grafo = {
            "A": ["B", "C"],
            "B": ["A", "D", "E"],
            "C": ["A", "F"],
            "D": ["B"],
            "E": ["B", "G"],
            "F": ["C", "G"],
            "G": ["E", "F"]
        }

    def mostrar_grafo(self):
        print("\n--- LISTA DE ADYACENCIA ---")
        for centro, conexiones in self.grafo.items():
            print(f"{centro}: {conexiones}")

    def bfs(self, inicio):
        if inicio not in self.grafo:
            return None

        visitados = {inicio}
        cola = deque([inicio])
        recorrido = []
        niveles = {inicio: 0}

        while cola:
            nodo = cola.popleft()
            recorrido.append(nodo)

            for vecino in self.grafo[nodo]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    niveles[vecino] = niveles[nodo] + 1
                    cola.append(vecino)

        return recorrido, niveles

    def dfs(self, inicio):
        if inicio not in self.grafo:
            return None

        visitados = set()
        recorrido = []

        def explorar(nodo):
            visitados.add(nodo)
            recorrido.append(nodo)

            for vecino in self.grafo[nodo]:
                if vecino not in visitados:
                    explorar(vecino)

        explorar(inicio)
        return recorrido

    def buscar_centro_bfs(self, inicio, destino):
        if inicio not in self.grafo or destino not in self.grafo:
            return None

        visitados = {inicio}
        cola = deque([(inicio, [inicio])])

        while cola:
            nodo, camino = cola.popleft()

            if nodo == destino:
                return camino

            for vecino in self.grafo[nodo]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    cola.append((vecino, camino + [vecino]))

        return None

    def agregar_centro(self, nuevo_centro, conexiones):
        if nuevo_centro not in self.grafo:
            self.grafo[nuevo_centro] = []

        for vecino in conexiones:
            if vecino not in self.grafo:
                self.grafo[vecino] = []

            if vecino not in self.grafo[nuevo_centro]:
                self.grafo[nuevo_centro].append(vecino)

            if nuevo_centro not in self.grafo[vecino]:
                self.grafo[vecino].append(nuevo_centro)


def programa_principal():
    red = RedDistribucion()

    while True:
        print("\n====================================")
        print("       RED DE DISTRIBUCIÓN")
        print("====================================")
        print("1. Mostrar grafo")
        print("2. Recorrido BFS")
        print("3. Recorrido DFS")
        print("4. Buscar un centro utilizando BFS")
        print("5. Agregar nuevo centro y conexiones")
        print("6. Salir")
        print("====================================")

        opcion = input("Seleccione una opción: ").upper()

        if opcion == "1":
            red.mostrar_grafo()

        elif opcion == "2":
            inicio = input("Ingrese el centro inicial: ").upper()

            if inicio not in red.grafo:
                print("Error: el centro ingresado no existe.")
            else:
                recorrido, niveles = red.bfs(inicio)

                print("\nRecorrido BFS:")
                print(" -> ".join(recorrido))

                print("\nNiveles:")
                for nivel in range(max(niveles.values()) + 1):
                    centros = [
                        centro for centro in recorrido
                        if niveles[centro] == nivel
                    ]
                    print(f"Nivel {nivel}: {', '.join(centros)}")

        elif opcion == "3":
            inicio = input("Ingrese el centro inicial: ").upper()

            if inicio not in red.grafo:
                print("Error: el centro ingresado no existe.")
            else:
                print("\nRecorrido DFS:")
                print(" -> ".join(red.dfs(inicio)))

        elif opcion == "4":
            inicio = input("Centro inicial: ").upper()
            destino = input("Centro que desea buscar: ").upper()

            camino = red.buscar_centro_bfs(inicio, destino)

            if camino is None:
                print("No existe una ruta hasta el centro solicitado.")
            else:
                print("\nCentro encontrado.")
                print("Ruta:", " -> ".join(camino))
                print("Cantidad de conexiones:", len(camino) - 1)

        elif opcion == "5":
            nuevo_centro = input("Ingrese el nuevo centro: ").upper()
            conexiones = []

            cantidad = int(
                input("¿Con cuántos centros desea conectar el nuevo centro?: ")
            )

            for i in range(cantidad):
                vecino = input(
                    f"Ingrese el centro para la conexión {i + 1}: "
                ).upper()

                if vecino in red.grafo:
                    conexiones.append(vecino)
                else:
                    print("Ese centro no existe.")

            red.agregar_centro(nuevo_centro, conexiones)

            print("\nCentro agregado correctamente.")
            red.mostrar_grafo()

        elif opcion == "6":
            print("\nPrograma finalizado.")
            break

        else:
            print("\nOpción no válida. Intente nuevamente.")


# Ejecuta el menú únicamente cuando lo necesites:
programa_principal()
```

#### Código base para benchmarking

```python
# ==============================
# PRUEBAS DE DESEMPEÑO
# ==============================

def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
    print("Generando datos ...", end="")
    results = []

    for size in range(1, n, step):
        red_prueba = RedDistribucion()
        red_prueba.grafo = {}

        # Se crea una red con la cantidad de centros indicada
        for i in range(size):
            centro = str(i)
            red_prueba.grafo[centro] = []

            if i > 0:
                red_prueba.grafo[centro].append(str(i - 1))
                red_prueba.grafo[str(i - 1)].append(centro)

        # Se mide el tiempo de BFS
        ti = time.perf_counter()
        resultado = red_prueba.bfs("0")
        tf = time.perf_counter()

        delta = (tf - ti) * 1000

        # Verifica que BFS haya recorrido todos los centros
        assert resultado is not None
        assert len(resultado[0]) == size

        print(".", end="")
        results.append((size, delta))

    print()
    return results


def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)

    plt.figure(figsize=dimensiones)
    plt.plot(x, y, marker="o")
    plt.xlabel("Cantidad de centros")
    plt.ylabel("Tiempo de ejecución (ms)")
    plt.title("Desempeño del algoritmo BFS")
    plt.grid(True)
    plt.show()


# Ejecución de las pruebas de desempeño
tiempos = bench(500, step=50)

x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]

dibujar(x, y)
```
