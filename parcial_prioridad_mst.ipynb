```python
import math
import matplotlib.pyplot as plt
 

lista_tareas = [
    [6, "T01", "Configurar servidor"],
    [10, "T02", "Reparar sensor"],
    [4, "T03", "Actualizar software"],
    [8, "T04", "Revisar camara"],
    [3, "T05", "Generar respaldo"],
    [7, "T06", "Configurar alarma"]
]
 
def max_heapify(arr, tam, idx):
    mas_grande = idx
    hijo_izq = 2 * idx + 1
    hijo_der = 2 * idx + 2
 
    if hijo_izq < tam and arr[hijo_izq][0] > arr[mas_grande][0]:
        mas_grande = hijo_izq
 
    if hijo_der < tam and arr[hijo_der][0] > arr[mas_grande][0]:
        mas_grande = hijo_der
 
    if mas_grande != idx:
        arr[idx], arr[mas_grande] = arr[mas_grande], arr[idx]
        max_heapify(arr, tam, mas_grande)
 
def build_max_heap(arr):
    tam = len(arr)
    for k in range(tam // 2 - 1, -1, -1):
        max_heapify(arr, tam, k)
 
def ver_arbol(arr):
    print("Arbol del heap:")
    niv = 0
    pos = 0
    while pos < len(arr):
        cant = 2 ** niv
        parte = arr[pos:pos + cant]
        linea = ""
        for t in parte:
            linea += f"{t[1]}(p={t[0]})  "
        print(f"Nivel {niv}: {linea}")
        pos += cant
        niv += 1
 
print("Inicial:", lista_tareas)
build_max_heap(lista_tareas)
print("Max-Heap:", lista_tareas)
ver_arbol(lista_tareas)
 

def extract_max(arr):
    if len(arr) == 0:
        return None
    if len(arr) == 1:
        return arr.pop()
    mayor_tarea = arr[0]
    arr[0] = arr.pop()
    max_heapify(arr, len(arr), 0)
    return mayor_tarea
 

def agregar_tarea(arr, tarea_nueva):
    arr.append(tarea_nueva)
    actual = len(arr) - 1
    while actual > 0:
        p = (actual - 1) // 2
        if arr[actual][0] > arr[p][0]:
            arr[actual], arr[p] = arr[p], arr[actual]
            actual = p
        else:
            break
 
sacada_1 = extract_max(lista_tareas)
print("\nProcesada 1:", sacada_1)
print("Heap ahora:", lista_tareas)
 
sacada_2 = extract_max(lista_tareas)
print("\nProcesada 2:", sacada_2)
print("Heap ahora:", lista_tareas)
 
agregar_tarea(lista_tareas, [12, "T07", "Falla en sistema contra incendios"])
print("\nCon T07:", lista_tareas)
ver_arbol(lista_tareas)
 

equipos = [
    ["A", "Servidor", 2, 2],
    ["B", "Camara", 7, 3],
    ["C", "Sensor", 4, 7],
    ["D", "Alarma", 10, 8],
    ["E", "Router", 13, 3],
    ["F", "Panel", 15, 7]
]
 
plt.figure(figsize=(7, 4))
for eq in equipos:
    plt.scatter(eq[2], eq[3], s=90)
    plt.text(eq[2], eq[3] + 0.3, f"{eq[0]} - {eq[1]}", ha="center", fontsize=9)
 
plt.title("Dispositivos")
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(0, 17)
plt.ylim(0, 10)
plt.grid(True)
plt.show()
 

conexiones = []
for i in range(len(equipos)):
    for j in range(i + 1, len(equipos)):
        id1, _, x1, y1 = equipos[i]
        id2, _, x2, y2 = equipos[j]
        d = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
        conexiones.append([d, id1, id2])
 
conexiones.sort()
 

lider = {}
for eq in equipos:
    lider[eq[0]] = eq[0]
 
def encontrar(x):
    while lider[x] != x:
        x = lider[x]
    return x
 
def unir(x, y):
    rx = encontrar(x)
    ry = encontrar(y)
    if rx == ry:
        return False
    lider[rx] = ry
    return True
 
mst = []
cable_total = 0.0
 
for d, u, v in conexiones:
    if unir(u, v):
        mst.append([u, v, d])
        cable_total += d
        if len(mst) == 5:
            break
 
print("\nMST:")
for u, v, d in mst:
    print(f"{u} - {v}: {d:.2f}")
print(f"Longitud total de cable: {cable_total:.2f}")
 

posiciones = {eq[0]: (eq[2], eq[3]) for eq in equipos}
etiquetas = {eq[0]: eq[1] for eq in equipos}
 
plt.figure(figsize=(8, 5))
for u, v, d in mst:
    x1, y1 = posiciones[u]
    x2, y2 = posiciones[v]
    plt.plot([x1, x2], [y1, y2], linewidth=2)
    plt.text((x1 + x2) / 2, (y1 + y2) / 2, f"{d:.2f}", fontsize=8)
 
for cod, (x, y) in posiciones.items():
    plt.scatter(x, y, s=110)
    plt.text(x, y + 0.3, f"{cod} ({etiquetas[cod]})", ha="center", fontsize=9)
 
plt.title(f"MST final - Cable: {cable_total:.2f}")
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(0, 17)
plt.ylim(0, 10)
plt.grid(True)
plt.show()
```
