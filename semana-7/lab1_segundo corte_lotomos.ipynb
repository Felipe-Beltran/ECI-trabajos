## 1) Descripción del problema **[1 pts]**
## Sistema de atención de una tienda

Una tienda desea desarrollar un pequeño sistema para controlar la atención de clientes, registrar las operaciones realizadas y administrar su catálogo de productos.



Para resolver el problema se deberán utilizar tres estructuras de datos diferentes:


Cola: para administrar los clientes que esperan ser atendidos.


Pila: para guardar el historial de las últimas operaciones realizadas.


Lista doblemente enlazada: para administrar y recorrer el catálogo de productos.


## Punto 1. Cola de clientes
Los clientes son atendidos en el mismo orden en que llegan a la tienda.

El usuario deberá ingresar inicialmente una cantidad n de clientes.

Ejemplo:


¿Cuántos clientes desea registrar?: 4 Ingrese el nombre del cliente 1: Ana Ingrese el nombre del cliente 2: Carlos Ingrese el nombre del cliente 3: Laura Ingrese el nombre del cliente 4: Mateo
La cola inicialmente sería:


Ana -> Carlos -> Laura -> Mateo


El programa deberá permitir:


Agregar un nuevo cliente a la cola.


Atender al siguiente cliente.


Mostrar los clientes que siguen esperando.


Por ejemplo:


Atendiendo a: Ana

Clientes esperando: Carlos -> Laura -> Mateo


Cuando llegue un nuevo cliente:


Ingrese el nuevo cliente: Sofía


La cola deberá quedar:


Carlos -> Laura -> Mateo -> Sofía


## Punto 2. Pila de operaciones


Cada vez que se atienda un cliente, su nombre deberá guardarse en una pila de clientes atendidos.


Por ejemplo, después de atender:


Ana Carlos Laura


La pila sería conceptualmente:



| Laura  | <- Tope | Carlos | | Ana    | ---------


El programa deberá permitir:

Agregar a la pila cada cliente que sea atendido.


Mostrar el último cliente atendido.


Deshacer la última atención.


Si se selecciona la opción:

Deshacer última atención
y Laura fue la última cliente atendida, deberá retirarse de la pila.


La pila quedaría:


| Carlos | <- Tope | Ana    | ---------


Además, Laura deberá regresar al inicio de la cola de clientes pendientes.



## Punto 3. Lista doblemente enlazada de productos


La tienda también necesita mantener un catálogo de productos.

Cada nodo deberá almacenar:

Código, Nombre, Precio.


Por ejemplo:

101 - Arroz - $4.500 

102 - Leche - $3.800 

103 - Pan - $2.500 


104 - Café - $12.000


Estos productos deberán almacenarse utilizando una lista doblemente enlazada.

Cada nodo tendrá:

anterior <- Producto -> siguiente


Ejemplo:


NULL <- Arroz <-> Leche <-> Pan <-> Café -> NULL

El programa deberá permitir:


Agregar un producto al final de la lista.


Mostrar todos los productos desde el primero hasta el último.


Mostrar los productos desde el último hasta el primero.


Buscar un producto utilizando su código.


Eliminar un producto utilizando su código.


Por ejemplo:


Ingrese el código que desea eliminar: 103


Después de eliminar el producto Pan, la lista deberá quedar:

NULL <- Arroz <-> Leche <-> Café -> NULL

## Pila de clientes atendidos

La pila  tiene una complejidad de O(1) en sus operaciones principales, tanto en el mejor como en el peor caso, porque cada cliente atendido se agrega directamente en el tope de la pila. De la misma manera, para mostrar el último cliente atendido o deshacer la última atención, solamente se debe acceder y retirar el elemento que está en el tope, sin recorrer los demás clientes. Por esto, push, pop y consultar el tope son O(1). Si se necesita mostrar todos los clientes de la pila, la complejidad sería O(n) porque se deben recorrer sus elementos.





**Graficas**





## Cola de clientes

La cola tiene una complejidad de O(1) en sus operaciones principales, tanto en el mejor como en el peor caso, porque los clientes se agregan al final y se atiende siempre al que está al inicio. Por ejemplo, agregar a Sofía o atender a Ana no requiere recorrer todos los clientes, sino acceder directamente al extremo correspondiente. Por eso, encolar y desencolar son O(1), independientemente de cuántos clientes haya en la cola. Mostrar todos los clientes que siguen esperando, en cambio, requiere recorrerlos, por lo que es O(n).




**Graficas**




## Lista doblemente enlazada de productos

La lista doblemente enlazada permite recorrer los productos tanto desde el primero hasta el último como desde el último hasta el primero gracias a los enlaces anterior y siguiente. Agregar un producto al final puede ser O(1) si se mantiene un puntero al último nodo, mientras que recorrer y mostrar todos los productos es O(n). Para buscar o eliminar un producto por su código, en el mejor caso puede ser O(1) si se encuentra inmediatamente, pero en el peor caso es O(n) porque se debe recorrer la lista hasta encontrarlo. La ventaja de que sea doblemente enlazada es que, una vez encontrado el nodo, eliminarlo requiere ajustar directamente sus enlaces anterior y siguiente.

```python
# ==========================================
# SISTEMA DE ATENCION DE UNA TIENDA
# ==========================================


# ==========================================
# NODO PARA LA LISTA DOBLEMENTE ENLAZADA
# ==========================================

class NodoProducto:

    def __init__(self, codigo, nombre, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.precio = precio
        self.anterior = None
        self.siguiente = None


# ==========================================
# LISTA DOBLEMENTE ENLAZADA
# ==========================================

class ListaProductos:

    def __init__(self):
        self.inicio = None
        self.fin = None

    def agregar(self, codigo, nombre, precio):

        nuevo = NodoProducto(codigo, nombre, precio)

        if self.inicio is None:
            self.inicio = nuevo
            self.fin = nuevo

        else:
            nuevo.anterior = self.fin
            self.fin.siguiente = nuevo
            self.fin = nuevo

    def mostrar_adelante(self):

        actual = self.inicio

        if actual is None:
            print("No hay productos")

        else:

            while actual is not None:

                print(
                    actual.codigo,
                    "-",
                    actual.nombre,
                    "- $",
                    actual.precio
                )

                actual = actual.siguiente

    def mostrar_atras(self):

        actual = self.fin

        if actual is None:
            print("No hay productos")

        else:

            while actual is not None:

                print(
                    actual.codigo,
                    "-",
                    actual.nombre,
                    "- $",
                    actual.precio
                )

                actual = actual.anterior

    def buscar(self, codigo):

        actual = self.inicio
        encontrado = None

        while actual is not None:

            if actual.codigo == codigo:

                encontrado = actual
                actual = None

            else:

                actual = actual.siguiente

        return encontrado

    def eliminar(self, codigo):

        actual = self.inicio
        eliminado = False

        while actual is not None and eliminado == False:

            if actual.codigo == codigo:

                if actual.anterior is None:

                    self.inicio = actual.siguiente

                    if self.inicio is not None:
                        self.inicio.anterior = None

                else:

                    actual.anterior.siguiente = actual.siguiente

                    if actual.siguiente is not None:
                        actual.siguiente.anterior = actual.anterior

                if actual == self.fin:
                    self.fin = actual.anterior

                eliminado = True

            else:

                actual = actual.siguiente

        return eliminado


# ==========================================
# COLA DE CLIENTES
# ==========================================

class ColaClientes:

    def __init__(self):
        self.clientes = []

    def agregar(self, nombre):

        self.clientes.append(nombre)

    def atender(self):

        if len(self.clientes) == 0:

            return None

        cliente = self.clientes.pop(0)

        return cliente

    def mostrar(self):

        if len(self.clientes) == 0:

            print("No hay clientes esperando")

        else:

            posicion = 0

            while posicion < len(self.clientes):

                if posicion > 0:
                    print("->", end=" ")

                print(self.clientes[posicion], end=" ")

                posicion += 1

            print()

    def devolver_al_inicio(self, nombre):

        self.clientes.insert(0, nombre)


# ==========================================
# PILA DE CLIENTES ATENDIDOS
# ==========================================

class PilaAtendidos:

    def __init__(self):
        self.clientes = []

    def agregar(self, nombre):

        self.clientes.append(nombre)

    def retirar(self):

        if len(self.clientes) == 0:

            return None

        cliente = self.clientes.pop()

        return cliente

    def ultimo(self):

        if len(self.clientes) == 0:

            return None

        return self.clientes[-1]

    def mostrar(self):

        if len(self.clientes) == 0:

            print("No hay clientes atendidos")

        else:

            posicion = len(self.clientes) - 1

            while posicion >= 0:

                if posicion == len(self.clientes) - 1:

                    print(
                        "|",
                        self.clientes[posicion],
                        "| <- Tope"
                    )

                else:

                    print(
                        "|",
                        self.clientes[posicion],
                        "|"
                    )

                posicion -= 1

            print("---------")


# ==========================================
# FUNCION PARA LEER NUMEROS
# ==========================================

def numero(texto):

    dato = input(texto)

    while dato.isdigit() == False:

        print("Escribe un numero")

        dato = input(texto)

    return int(dato)


# ==========================================
# PROGRAMA PRINCIPAL
# ==========================================

def ejecutar_programa():

    # ==========================================
    # CREAR ESTRUCTURAS
    # ==========================================

    cola = ColaClientes()

    pila = PilaAtendidos()

    productos = ListaProductos()


    # ==========================================
    # REGISTRO INICIAL DE CLIENTES
    # ==========================================

    cantidad = numero(
        "¿Cuantos clientes desea registrar?: "
    )

    contador = 1

    while contador <= cantidad:

        nombre = input(
            "Ingrese el nombre del cliente "
            + str(contador)
            + ": "
        )

        cola.agregar(nombre)

        contador += 1


    # ==========================================
    # MENU PRINCIPAL
    # ==========================================

    opcion = -1

    while opcion != 0:

        print()
        print("========== MENU ==========")
        print("1. Agregar nuevo cliente")
        print("2. Atender siguiente cliente")
        print("3. Mostrar clientes esperando")
        print("4. Mostrar ultimo cliente atendido")
        print("5. Deshacer ultima atencion")
        print("6. Mostrar pila de clientes atendidos")
        print("7. Agregar producto")
        print("8. Mostrar productos de primero a ultimo")
        print("9. Mostrar productos de ultimo a primero")
        print("10. Buscar producto")
        print("11. Eliminar producto")
        print("0. Salir")
        print("==========================")

        opcion = numero("Opcion: ")


        if opcion == 1:

            nombre = input(
                "Ingrese el nuevo cliente: "
            )

            cola.agregar(nombre)

            print(
                "Cliente agregado correctamente."
            )


        elif opcion == 2:

            cliente = cola.atender()

            if cliente is None:

                print(
                    "No hay clientes para atender."
                )

            else:

                pila.agregar(cliente)

                print(
                    "Atendiendo a:",
                    cliente
                )

                print(
                    "Clientes esperando:"
                )

                cola.mostrar()


        elif opcion == 3:

            print(
                "Clientes esperando:"
            )

            cola.mostrar()


        elif opcion == 4:

            cliente = pila.ultimo()

            if cliente is None:

                print(
                    "No hay clientes atendidos."
                )

            else:

                print(
                    "Ultimo cliente atendido:",
                    cliente
                )


        elif opcion == 5:

            cliente = pila.retirar()

            if cliente is None:

                print(
                    "No hay atenciones para deshacer."
                )

            else:

                cola.devolver_al_inicio(
                    cliente
                )

                print(
                    "Se deshizo la atencion de:",
                    cliente
                )

                print(
                    "El cliente regreso al inicio de la cola."
                )


        elif opcion == 6:

            print(
                "Pila de clientes atendidos:"
            )

            pila.mostrar()


        elif opcion == 7:

            codigo = numero("Codigo: ")

            nombre = input(
                "Nombre del producto: "
            )

            precio = numero("Precio: ")

            productos.agregar(
                codigo,
                nombre,
                precio
            )

            print(
                "Producto agregado correctamente."
            )


        elif opcion == 8:

            print(
                "Catalogo de primero a ultimo:"
            )

            productos.mostrar_adelante()


        elif opcion == 9:

            print(
                "Catalogo de ultimo a primero:"
            )

            productos.mostrar_atras()


        elif opcion == 10:

            codigo = numero(
                "Ingrese el codigo que desea buscar: "
            )

            producto = productos.buscar(
                codigo
            )

            if producto is None:

                print(
                    "Producto no encontrado."
                )

            else:

                print(
                    producto.codigo,
                    "-",
                    producto.nombre,
                    "- $",
                    producto.precio
                )


        elif opcion == 11:

            codigo = numero(
                "Ingrese el codigo que desea eliminar: "
            )

            eliminado = productos.eliminar(
                codigo
            )

            if eliminado:

                print(
                    "Producto eliminado correctamente."
                )

            else:

                print(
                    "Producto no encontrado."
                )


        elif opcion == 0:

            print(
                "Programa finalizado."
            )


        else:

            print(
                "Opcion no valida."
            )
```

```python
# Calcula el tiempo de ejecucion con base a diferentes tamaños de la entrada
# y nos da una lista de resultados, cada resultado contiene el tamaño de la entrada
# y el tiempo de ejecucion
# TENER EN CUENTA ESTAS ENTRADAS PUEDEN VARIAR SEGÚN EL PROBLEMA
def prueba_rendimiento(size):

    cola_prueba = ColaClientes()

    contador = 0

    while contador < size:

        cola_prueba.agregar("Cliente")

        contador += 1

    cliente = cola_prueba.atender()

    return cliente

def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
    print("Generando datos ...", end='')
    from random import randint
    import time
    results: list[tuple[int, float]] = []


    # AJUSTAR LAS ENTRADAS SEGÚN EL PROBLEMA
    for size in range(1, n, step):
        input = size

        ti = time.perf_counter()
        output = prueba_rendimiento(input)
        tf = time.perf_counter()


        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
    return results


# === CODIGO PARA DIBUJAR ===


def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    import matplotlib.pyplot as plt
    plt.figure(figsize=dimensiones)
    plt.plot(x, y)
    plt.show()


# === LLAMADO ===
tiempos = bench(10, step=1)


x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]


dibujar(x, y)
```
