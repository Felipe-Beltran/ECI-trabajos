## 7) Análisis de complejidad **[15 pts]**


El análisis de costo permite determinar cuántas veces se ejecuta cada instrucción del algoritmo Cambio de Monedas. El ciclo for recorre los k tipos de monedas, por lo que se ejecuta k+1 veces contando la condición de finalización. Las operaciones internas dependen de la cantidad de monedas utilizadas, representada por sumatoria t_i. Por esta razón, el costo total depende principalmente de k y de la cantidad de monedas generadas.

Mejor caso

El mejor caso ocurre cuando se utilizan pocas monedas o cuando para algunos tipos de moneda se cumple num = 0. En este caso, las instrucciones de las líneas 8, 9 y 10 se ejecutan menos veces.

T_mejor(k) = c₁ + c₂ + c₃ + c₄ + c₅(k + 1) + c₆k + c₇k + c₈Σtᵢ* + c₉Σtᵢ* + c₁₀Σtᵢ* + c₁₁

por lo tanto:


T_mejor(k) = Ak + Bk

la funcion:


T_mejor(n) = O(n)


Peor caso

El peor caso ocurre cuando se utilizan muchas monedas y las instrucciones internas se ejecutan repetidamente para cada tipo de moneda. En este caso, sumatoria t_i es grande.

T_peor(k) = c₁ + c₂ + c₃ + c₄ + c₅(k + 1) + c₆k + c₇k+ c₈Σtᵢ + c₉Σtᵢ + c₁₀Σtᵢ + c₁₁

Por lo tanto:

T_peor(k) = Ak + BΣtᵢ + D

 la funcion:

 T_peor(n) = O(n)

Conclusión

Aunque el número de operaciones cambia entre el mejor y el peor caso, ambos mantienen un crecimiento lineal respecto a la cantidad de monedas procesadas. Por eso, la complejidad temporal del algoritmo es:

Mejor caso → T(n) = O(n)


Peor caso  → T(n) = O(n)



## 8) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
monedas = [1000, 500, 200, 100, 50]

while True:
    valor = input("Ingrese el valor del cambio: ").strip()

    if not valor.isdigit():
        print("Ingrese un valor numérico válido.")
        continue

    cambio = int(valor)

    if cambio <= 0:
        print("El monto debe ser mayor a 0.")
        continue

    if cambio % 50 != 0:
        print("No se puede dar el cambio exacto con monedas de 50 o más.")
        continue

    break

original = cambio
total_monedas = 0
resultado = []

for m in monedas:
    num = cambio // m
    if num > 0:
        total_monedas += num
        cambio %= m
        resultado.extend([str(m)] * num)

print("\nUna posible solución sería:")
print(" + ".join(resultado) + " = " + str(original))
print("Por lo tanto:")
print("Cantidad mínima de monedas:", total_monedas)


monedas = [1000, 500, 200, 100, 50]

# --- Función para calcular el cambio ---
def cambio_minimo(valor, monedas=monedas):
    resultado = []
    total_monedas = 0
    for m in monedas:
        num = valor // m
        if num > 0:
            total_monedas += num
            valor %= m
            resultado.extend([m] * num)
    return resultado, total_monedas



```

## 9) pruebas **[15 pts]**

```python
def test(input, expected_output):
    got_output = cambio_minimo(input)
    assert got_output == expected_output, f"Se obtuvo {got_output}, pero se esperaba {expected_output}"
    print(f"✅ Caso de prueba {input} -> {expected_output} pasó correctamente")
```

#### Código base para benchmarking

```python
# Calcula el tiempo de ejecucion con base a diferentes tamaños de la entrada
# y nos da una lista de resultados, cada resultado contiene el tamaño de la entrada
# y el tiempo de ejecucion
# TENER EN CUENTA ESTAS ENTRADAS PUEDEN VARIAR SEGÚN EL PROBLEMA


def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
    print("Generando datos ...", end='')
    from random import randint
    import time
    results: list[tuple[int, float]] = []


    # AJUSTAR LAS ENTRADAS SEGÚN EL PROBLEMA
    for size in range(1, n, step):
        input = size

        ti = time.perf_counter()
        output = cambio_minimo(input)
        tf = time.perf_counter()


        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
    return results


# === CODIGO PARA DIBUJAR ===


def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    import matplotlib.pyplot as plt
    plt.figure(figsize=dimensiones)
    plt.plot(x, y)
    plt.show()


# === LLAMADO ===
tiempos = bench(10, step=1)


x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]


dibujar(x, y)
```
