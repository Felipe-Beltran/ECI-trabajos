### 5) Análisis de complejidad **[15 pts]**

## 1. Árbol Binario de Búsqueda (BST)

El Árbol Binario de Búsqueda construye el árbol insertando los n elementos uno por uno y comparando cada nuevo valor con los nodos existentes para decidir si debe ir a la izquierda o a la derecha. La complejidad depende de la altura que alcance el árbol. En el mejor caso, cuando el árbol queda aproximadamente balanceado, la altura es proporcional a log(n), por lo que las inserciones tienen un costo logarítmico y, al realizarse n inserciones, la función de complejidad es T(n)=n log(n), dando O(n log n). En el peor caso, el árbol puede quedar completamente desbalanceado y su altura puede llegar a n, haciendo que las inserciones tengan un costo creciente hasta O(n). Al realizar las n inserciones, la función queda T(n)=n^2, por lo que el peor caso es O(n^2).

Función de complejidad:

T(n)=nlog(n)


Mejor caso:

T(n)=n log(n)

Peor caso:

O(n^2)

## 2. Árbol AVL

El Árbol AVL construye el árbol utilizando los n elementos y, después de cada inserción, calcula el factor de balance para determinar si es necesario realizar una rotación. Estas rotaciones permiten mantener el árbol balanceado y conservar una altura proporcional a \log(n). Por esta razón, cada inserción tiene una complejidad O(\log n), tanto en el mejor como en el peor caso. Como el ejercicio realiza n inserciones para construir el árbol completo, la función de complejidad es T(n)=n\log(n). Por lo tanto, el mejor caso es O(n\log n) y el peor caso también es O(n\log n), ya que el balanceo evita que el árbol alcance una altura lineal.

Función de complejidad:

T(n)=n log(n)

Mejor caso:

T(n)=n log(n)

Peor caso:

T(n)=n log(n)

```python
# ==========================================================
# 1. ESTRUCTURAS DE LOS NODOS
# ==========================================================

def crear_nodo(valor):
    """Crea un nodo básico para un BST."""
    return {"valor": valor, "izq": None, "der": None}


def crear_nodo_avl(valor):
    """Crea un nodo para un AVL (incluye altura)."""
    return {"valor": valor, "izq": None, "der": None, "altura": 1}


# ==========================================================
# 2. FUNCIONES PARA BST (Árbol Binario de Búsqueda)
# ==========================================================

def insertar_bst(nodo, valor):
    if nodo is None:
        return crear_nodo(valor)

    if valor < nodo["valor"]:
        nodo["izq"] = insertar_bst(nodo["izq"], valor)
    elif valor > nodo["valor"]:
        nodo["der"] = insertar_bst(nodo["der"], valor)

    return nodo


# ==========================================================
# 3. FUNCIONES PARA AVL (Árbol Auto-balanceado)
# ==========================================================

def get_altura(nodo):
    return nodo["altura"] if nodo else 0


def get_balance(nodo):
    return get_altura(nodo["izq"]) - get_altura(nodo["der"]) if nodo else 0


def rotar_derecha(y):
    x = y["izq"]
    temp = x["der"]

    x["der"] = y
    y["izq"] = temp

    y["altura"] = 1 + max(get_altura(y["izq"]), get_altura(y["der"]))
    x["altura"] = 1 + max(get_altura(x["izq"]), get_altura(x["der"]))
    return x


def rotar_izquierda(x):
    y = x["der"]
    temp = y["izq"]

    y["izq"] = x
    x["der"] = temp

    x["altura"] = 1 + max(get_altura(x["izq"]), get_altura(x["der"]))
    y["altura"] = 1 + max(get_altura(y["izq"]), get_altura(y["der"]))
    return y


def insertar_avl(nodo, valor):
    if nodo is None:
        return crear_nodo_avl(valor)

    if valor < nodo["valor"]:
        nodo["izq"] = insertar_avl(nodo["izq"], valor)
    elif valor > nodo["valor"]:
        nodo["der"] = insertar_avl(nodo["der"], valor)
    else:
        return nodo  # No duplicados

    # Actualizar altura del nodo actual
    nodo["altura"] = 1 + max(get_altura(nodo["izq"]), get_altura(nodo["der"]))

    # Verificar si se desbalanceó
    balance = get_balance(nodo)

    # Caso Izquierda-Izquierda
    if balance > 1 and valor < nodo["izq"]["valor"]:
        return rotar_derecha(nodo)

    # Caso Derecha-Derecha
    if balance < -1 and valor > nodo["der"]["valor"]:
        return rotar_izquierda(nodo)

    # Caso Izquierda-Derecha
    if balance > 1 and valor > nodo["izq"]["valor"]:
        nodo["izq"] = rotar_izquierda(nodo["izq"])
        return rotar_derecha(nodo)

    # Caso Derecha-Izquierda
    if balance < -1 and valor < nodo["der"]["valor"]:
        nodo["der"] = rotar_derecha(nodo["der"])
        return rotar_izquierda(nodo)

    return nodo


# ==========================================================
# 4. RECORRIDOS (Sirven para BST y AVL)
# ==========================================================

def preorden(nodo):
    if nodo is None:
        return []
    return [nodo["valor"]] + preorden(nodo["izq"]) + preorden(nodo["der"])


def inorden(nodo):
    if nodo is None:
        return []
    return inorden(nodo["izq"]) + [nodo["valor"]] + inorden(nodo["der"])


def postorden(nodo):
    if nodo is None:
        return []
    return postorden(nodo["izq"]) + postorden(nodo["der"]) + [nodo["valor"]]


# ==========================================================
# 5. FUNCIONES PRINCIPALES PARA TUS PRUEBAS
# ==========================================================

def bst_preorden(datos):
    raiz = None
    for x in datos:
        raiz = insertar_bst(raiz, x)
    return preorden(raiz)


def bst_inorden(datos):
    raiz = None
    for x in datos:
        raiz = insertar_bst(raiz, x)
    return inorden(raiz)


def bst_postorden(datos):
    raiz = None
    for x in datos:
        raiz = insertar_bst(raiz, x)
    return postorden(raiz)


def avl_preorden(datos):
    raiz = None
    for x in datos:
        raiz = insertar_avl(raiz, x)
    return preorden(raiz)


def avl_inorden(datos):
    raiz = None
    for x in datos:
        raiz = insertar_avl(raiz, x)
    return inorden(raiz)


def avl_postorden(datos):
    raiz = None
    for x in datos:
        raiz = insertar_avl(raiz, x)
    return postorden(raiz)


```
