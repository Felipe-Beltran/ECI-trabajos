## 7) Análisis de complejidad **[15 pts]**
Incluye:
1. Merge Sort

Merge Sort divide la lista en dos partes, ordena cada mitad y luego las combina de forma ordenada.

Ecuación de recurrencia:

T(n) = 2T(n/2) + n

Complejidad:

T(n) = O(n log n)

Mejor caso: O(n log n)

Caso promedio: O(n log n)

Peor caso: O(n log n)



2.Búsqueda Binaria

La Búsqueda Binaria se realiza sobre la lista previamente ordenada. En cada comparación, se descarta aproximadamente la mitad de los elementos.

Ecuación de recurrencia:

T(n) = T(n/2) + 1

Complejidad:

T(n) = O(log n)

Mejor caso: O(1)

Caso promedio: O(log n)

Peor caso: O(log n)





El mejor caso ocurre en la búsqueda binaria cuando el producto que se está buscando se encuentra exactamente en la posición central de la lista ordenada.
El peor caso ocurre cuando el producto no está registrado en la lista o cuando se necesita realizar la mayor cantidad de divisiones posibles para encontrarlo.


## 8) Código funcional **[15 pts]**



```python
continuar = "s"

while continuar == "s" or continuar == "S":
    print("\n--- REGISTRO Y BÚSQUEDA DE PRODUCTOS ---")

   
    entrada = input("Ingrese la cantidad de productos: ")
    while not entrada.isdigit() or int(entrada) <= 0:
        print("Error: Ingrese un número entero positivo (sin letras ni espacios).")
        entrada = input("Ingrese la cantidad de productos nuevamente: ")
    
    n = int(entrada)
    productos = [0] * n

    
    print("\nNota: Escriba los números juntos sin espacios (ejemplo: 12).")
    for i in range(n):
        val = input(f"Ingrese el codigo del producto {i+1}: ")
        while not val.isdigit():
            print("Error: El código debe ser solo números enteros (sin letras ni espacios).")
            val = input(f"Ingrese el codigo del producto {i+1} nuevamente: ")
        productos[i] = int(val)

    print("\nLista original:")
    print(productos)

   
    for i in range(n - 1):
        for j in range(i + 1, n):
            if productos[i] > productos[j]:
                aux = productos[i]
                productos[i] = productos[j]
                productos[j] = aux

    print("\nLista ordenada:")
    print(productos)

    
    busqueda = input("\nIngrese el codigo que desea buscar: ")
    while not busqueda.isdigit():
        print("Error: Ingrese solo números para el código de búsqueda.")
        busqueda = input("Ingrese el codigo que desea buscar nuevamente: ")
    
    codigo = int(busqueda)

    inicio = 0
    final = n - 1
    posicion = -1

    while inicio <= final and posicion == -1:
        medio = (inicio + final) // 2

        if productos[medio] == codigo:
            posicion = medio
        elif codigo < productos[medio]:
            final = medio - 1
        else:
            inicio = medio + 1

    if posicion != -1:
        print("El producto fue encontrado.")
        print("Se encuentra en la posicion:", posicion)
    else:
        print("El producto no se encuentra registrado.")

    continuar = input("\n¿Desea realizar otra operación? (s/n): ")

print("\n¡Programa finalizado!")
```
